## Description
> ##### The Hangman program randomly selects a secret word from a list of secret words.
> ##### The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to guess the fruit name correctly.
> ##### When a letter in that word is guessed correctly, that letter position in the word is made visible in the output.
> ##### In this way, all letters of the word are to be guessed before all the chances are over.
> ##### For convenience, the number of chances equals the word length + 2.
> ##### For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five letter word.
> ##### Note : Total number of Guesses either right or wrong equals word length + 2.
> ##### Make game repeats until player ends the game.
> ##### Bonus : Create a function that takes 3 parameters `Easy`, `Medium` and `Hard` where `Medium` has more difficult words to be guessed than `Easy` and so on.

```python
import random

def choose_word(difficulty):
    """Return a random fruit based on the selected difficulty."""
    words = {
        "easy": ["fig", "kiwi", "lime", "pear", "plum", "date"],
        "medium": ["mango", "grape", "lemon", "peach", "melon", "guava"],
        "hard": ["pineapple", "pomegranate", "watermelon", "strawberry",
                 "blueberry", "cantaloupe"]
    }

    difficulty = difficulty.lower()
    if difficulty not in words:
        print("Invalid difficulty. Medium will be selected.")
        difficulty = "medium"

    return random.choice(words[difficulty])


def play_hangman():
    print("Welcome to Hangman!")

    while True:
        difficulty = input("Choose difficulty (Easy, Medium, Hard): ").strip().lower()
        secret_word = choose_word(difficulty)

        guessed_letters = set()
        remaining_guesses = len(secret_word) + 2

        print(f"\nGuess the fruit! You have {remaining_guesses} guesses.")

        while remaining_guesses > 0:
            # Display guessed letters and hide the remaining ones.
            displayed_word = " ".join(
                letter if letter in guessed_letters else "_"
                for letter in secret_word
            )
            print("\nWord:", displayed_word)
            print("Guesses remaining:", remaining_guesses)

            if all(letter in guessed_letters for letter in secret_word):
                print(f"Congratulations! You guessed the word: {secret_word}")
                break

            guess = input("Guess a letter: ").strip().lower()

            if len(guess) != 1 or not guess.isalpha():
                print("Please enter one alphabetic letter.")
                continue

            if guess in guessed_letters:
                print("You already guessed that letter. Try another one.")
                continue

            guessed_letters.add(guess)

            if guess in secret_word:
                print("Correct guess!")
            else:
                print("Wrong guess!")
                remaining_guesses -= 1

        else:
            print(f"\nGame over! The fruit was: {secret_word}")

        again = input("\nWould you like to play again? (yes/no): ").strip().lower()
        if again not in ("yes", "y"):
            print("Thanks for playing Hangman!")
            break


# Start the game
play_hangman()
```
